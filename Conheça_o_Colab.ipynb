```
def taxa(quantidade):

    if quantidade < 15:
        return 20.00
    elif quantidade == 15:
        return 14.00
    else:
        return 12.00


def cadastro():

    hospedes = {}
    for i in range(5):
        while True:
            nome = input(f"Digite o nome do {i+1}º hóspede: ").strip().title()
            if nome in hospedes:
                print(" Nome já cadastrado! Digite um nome diferente.")
            else:
                break


        while True:
            try:
                diarias = int(input(f"Digite a quantidade de diárias de {nome}: "))
                if diarias <= 0:
                    print(" Quantidade inválida. Deve ser maior que 0.")
                else:
                    break
            except ValueError:
                print(" Digite apenas números inteiros.")

        hospedes[nome] = diarias

    return hospedes


def pesquisar(hospedes):

    nome = input("Digite o nome do hóspede para pesquisa: ").strip().title()

    if nome in hospedes:
        print(f"\nO hóspede {nome} já se hospedou {hospedes[nome]} vez(es).")
    else:
        print(f"\nO hóspede {nome} ainda não está no sistema.")
        # Cadastra o novo hóspede
        while True:
            try:
                diarias = int(input(f"Digite a quantidade de diárias de {nome}: "))
                if diarias <= 0:
                    print("⚠️ Quantidade inválida. Deve ser maior que 0.")
                else:
                    break
            except ValueError:
                print("⚠️ Digite apenas números inteiros.")

        hospedes[nome] = diarias


        valor_diaria = 300
        taxa_servico = taxa(diarias)
        total = valor_diaria * diarias + taxa_servico

        print(f"\nHóspede {nome} adicionado com sucesso!")
        print(f"Valor total a pagar: R$ {total:.2f} (inclui taxa de R$ {taxa_servico:.2f})")



print("🏨 Sistema de Cadastro de Hóspedes 🏨\n")

hospedes = cadastro()
pesquisar(hospedes)
```
