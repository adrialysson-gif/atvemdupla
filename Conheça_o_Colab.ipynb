```




livros = []

def adicionar_livro():
    print("\n=== Adicionar Livro ===")
    titulo = input("Digite o título do livro: ").strip()
    autor = input("Digite o nome do autor: ").strip()


    if titulo == "" or autor == "":
        print("Você precisa digitar o título e o autor!\n")
        return


    livros.append({
        "titulo": titulo,
        "autor": autor,
        "emprestado": False
    })

    print(f"O livro '{titulo}' foi adicionado com sucesso!\n")


def emprestar_livro():
    print("\n=== Emprestar Livro ===")
    titulo = input("Digite o título do livro que deseja emprestar: ").strip()

    for livro in livros:
        if livro["titulo"].lower() == titulo.lower():
            if livro["emprestado"]:
                print("Esse livro já está emprestado!\n")
            else:
                livro["emprestado"] = True
                print(f"O livro '{titulo}' foi emprestado com sucesso!\n")
            return

    print("Livro não encontrado no acervo.\n")


def devolver_livro():
    print("\n=== Devolver Livro ===")
    titulo = input("Digite o título do livro que deseja devolver: ").strip()

    for livro in livros:
        if livro["titulo"].lower() == titulo.lower():
            if not livro["emprestado"]:
                print("Esse livro já estava disponível!\n")
            else:
                livro["emprestado"] = False
                print(f"O livro '{titulo}' foi devolvido com sucesso!\n")
            return

    print("Livro não encontrado no acervo.\n")


def listar_livros_disponiveis():
    print("\n=== Livros Disponíveis ===")
    encontrou = False

    for livro in livros:
        if not livro["emprestado"]:
            print(f"• {livro['titulo']} ({livro['autor']})")
            encontrou = True

    if not encontrou:
        print("Nenhum livro disponível no momento.")
    print()


def buscar_livro():
    print("\n=== Buscar Livro ===")
    titulo = input("Digite o título do livro que deseja buscar: ").strip()

    for livro in livros:
        if livro["titulo"].lower() == titulo.lower():
            status = "Emprestado" if livro["emprestado"] else "Disponível"
            print(f"\nTítulo: {livro['titulo']}")
            print(f"Autor: {livro['autor']}")
            print(f"Status: {status}\n")
            return

    print("Livro não encontrado.\n")


# Menu principal
def menu():
    while True:
        print("==== Biblioteca Comunitária ====")
        print("1 - Adicionar livro")
        print("2 - Emprestar livro")
        print("3 - Devolver livro")
        print("4 - Listar livros disponíveis")
        print("5 - Buscar livro")
        print("6 - Sair")

        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            adicionar_livro()
        elif opcao == "2":
            emprestar_livro()
        elif opcao == "3":
            devolver_livro()
        elif opcao == "4":
            listar_livros_disponiveis()
        elif opcao == "5":
            buscar_livro()
        elif opcao == "6":
            print("Saindo do sistema... Até logo!")
            break
        else:
            print("Opção inválida! Tente novamente.\n")



menu()

```

```
veiculos = []
servicos = []


def registrar_veiculo():
    print("\n=== Registrar Veículo ===")
    placa = input("Digite a placa do veículo: ").strip().upper()
    modelo = input("Digite o modelo do veículo: ").strip()
    dono = input("Digite o nome do dono: ").strip()

    if placa == "" or modelo == "" or dono == "":
        print(" Todos os campos devem ser preenchidos!\n")
        return


    for v in veiculos:
        if v["placa"] == placa:
            print(" Já existe um veículo com essa placa!\n")
            return

    veiculos.append({
        "placa": placa,
        "modelo": modelo,
        "dono": dono
    })
    print(f" Veículo {modelo} com placa {placa} registrado com sucesso!\n")


def registrar_servico():
    print("\n=== Registrar Serviço ===")
    placa = input("Digite a placa do veículo: ").strip().upper()


    for v in veiculos:
        if v["placa"] == placa:
            descricao = input("Digite a descrição do serviço: ").strip()
            valor = input("Digite o valor do serviço (em R$): ").strip()

            if descricao == "" or valor == "":
                print(" Todos os campos devem ser preenchidos!\n")
                return

            servicos.append({
                "placa": placa,
                "descricao": descricao,
                "valor": valor
            })
            print(f" Serviço '{descricao}' registrado para o veículo {placa}!\n")
            return

    print(" Veículo não encontrado! Cadastre-o antes de registrar o serviço.\n")


def listar_veiculos():
    print("\n=== Lista de Veículos Cadastrados ===")
    if len(veiculos) == 0:
        print("Nenhum veículo cadastrado ainda.\n")
    else:
        for v in veiculos:
            print(f"Placa: {v['placa']} | Modelo: {v['modelo']} | Dono: {v['dono']}")
        print()


def listar_servicos():
    print("\n=== Lista de Serviços Realizados ===")
    if len(servicos) == 0:
        print("Nenhum serviço registrado ainda.\n")
    else:
        for s in servicos:
            print(f"Placa: {s['placa']} | Serviço: {s['descricao']} | Valor: R$ {s['valor']}")
        print()


def buscar_veiculo():
    print("\n=== Buscar Veículo ===")
    placa = input("Digite a placa do veículo: ").strip().upper()

    for v in veiculos:
        if v["placa"] == placa:
            print(f"\nPlaca: {v['placa']}")
            print(f"Modelo: {v['modelo']}")
            print(f"Dono: {v['dono']}\n")
            return

    print(" Veículo não encontrado.\n")


def menu():
    while True:
        print("==== Oficina Mecânica ====")
        print("1 - Registrar veículo")
        print("2 - Registrar serviço")
        print("3 - Listar veículos")
        print("4 - Listar serviços")
        print("5 - Buscar veículo")
        print("6 - Sair")

        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            registrar_veiculo()
        elif opcao == "2":
            registrar_servico()
        elif opcao == "3":
            listar_veiculos()
        elif opcao == "4":
            listar_servicos()
        elif opcao == "5":
            buscar_veiculo()
        elif opcao == "6":
            print("Encerrando o sistema... Até logo! ")
            break
        else:
            print("Opção inválida! Tente novamente.\n")



menu()
```
